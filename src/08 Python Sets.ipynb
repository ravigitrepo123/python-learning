####Python Sets
Sets are used to store multiple items in a single variable.

Set is one of 4 built-in data types in Python used to store collections of data.

Sets are written with curly brackets.

```
set1 = {"clever","studies","classes"}
print(set1)
```

####Set Items
Set items are unordered, unchangeable, and do not allow duplicate values.

####Unordered
Unordered means that the items in a set do not have a defined order.

Set items can appear in a different order every time you use them, and cannot be referred to by index or key

####Unchangeable
Set items are unchangeable, meaning that we cannot change the items after the set has been created.

Once a set is created, you cannot change its items, but you can remove items and add new items.

####Duplicates Not Allowed
Sets cannot have two items with the same value.

```
#True and 1 is considered the same value:
set1 = {"clever","studies","classes",1,2,True}
print(set1)
```

```
set1 = {"clever","studies","classes",0,False,True}
print(set1)
```

####Get the Length of a Set
To determine how many items a set has, use the len() function.

```
set1 = {"clever","studies","classes"}
print(len(set1))
```

```
#Set items can be of any data type:
set1 = {"clever","studies","classes"}
set2 = {1, 5, 7, 9, 3}
set3 = {True, False, False}
print(set1)
print(set2)
print(set3)
```

```
set1 = {"clever","studies","classes"}
print(type(set1))
```

####The set() Constructor
It is also possible to use the set() constructor to make a set.

```
set1 = set(("clever","studies","classes"))
print(set1)
```

####Access Items
You cannot access items in a set by referring to an index or a key.

But you can loop through the set items using a for loop, or ask if a specified value is present in a set, by using the in keyword.

```
set1 = {"clever","studies","classes"}
for x in set1:
  print(x)
```

```
set1 = {"clever","studies","classes"}
print("studies" in set1)
```

```
set1 = {"clever","studies","classes"}
print("studies" not in set1)
```

####Add Items
Once a set is created, you cannot change its items, but you can add new items.

To add one item to a set use the add() method.

```
set1 = {"clever","studies","classes"}
set1.add("are going on")
print(set1)
```

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set1.update(set2)
print(set1)
```

####Add Any Iterable
The object in the update() method does not have to be a set, it can be any iterable object (tuples, lists, dictionaries etc.)

```
set1 = {"clever","studies","classes"}
set2 = ["are","going","on"]
set1.update(set2)
print(set1)
```

####Remove Item
To remove an item in a set, use the remove(), or the discard() method.

```
set1 = {"clever","studies","classes"}
set1.remove("classes")
print(set1)
```

```
set1 = {"clever","studies","classes"}
set1.discard("classes")
print(set1)
```

```
set1 = {"clever","studies","classes"}
x=set1.pop()
print(x)
print(set1)
#Sets are unordered, so when using the pop() method, you do not know which item that gets removed.
```

```
set1 = {"clever","studies","classes"}
set1.clear()
print(set1)
```

```
set1 = {"clever","studies","classes"}
del set1
print(set1)
```

####Union
The union() method returns a new set with all items from both sets.

```
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)
```

```
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1 | set2
print(set3)
```

####Join Multiple Sets
All the joining methods and operators can be used to join multiple sets.

When using a method, just add more sets in the parentheses, separated by commas:

```
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"clever","studies","classes"}
set4 = {"are","going","on"}
myset = set1.union(set2, set3, set4)
print(myset)
```

```
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"clever","studies","classes"}
set4 = {"are","going","on"}
myset = set1 | set2 |set3 | set4
print(myset)
```

####Join a Set and a Tuple
The union() method allows you to join a set with other data types, like lists or tuples.

The result will be a set.

```
x = {"a", "b", "c"}
y = (1, 2, 3)
z = x.union(y)
print(z)
#The  | operator only allows you to join sets with sets, and not with other data types like you can with the  union() method.
```

####Update
The update() method inserts all items from one set into another.

The update() changes the original set, and does not return a new set.

```
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set1.update(set2)
print(set1)
```

####Intersection
Keep ONLY the duplicates

The intersection() method will return a new set, that only contains the items that are present in both sets.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.intersection(set2)
print(set3)
```

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1 & set2
print(set3)
#The & operator only allows you to join sets with sets, and not with other data types like you can with the intersection() method.
```

####The intersection_update() 
This method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set1.intersection_update(set2)
print(set1)
```

```
set1 = {"clever",1,"studies",1,"classes",0}
set2 = {True,"are","going",False,"on",2}
set3 = set1.intersection(set2)
print(set3)
```

####Difference
The difference() method will return a new set that will contain only the items from the first set that are not present in the other set.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.difference(set2)
print(set3)
```

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1 - set2
print(set3)
#The - operator only allows you to join sets with sets, and not with other data types like you can with the difference() method
```

####Difference_update() method 
This will also keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.

```
#Use the difference_update() method to keep the items that are not present in both sets:
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set1.difference_update(set2)
print(set1)
```

####Symmetric Differences
The symmetric_difference() method will keep only the elements that are NOT present in both sets.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.symmetric_difference(set2)
print(set3)
```

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1 ^ set2
print(set3)
#The ^ operator only allows you to join sets with sets, and not with other data types like you can with the symmetric_difference() method.
```

####Symmetric_difference_update() method
This will also keep all but the duplicates, but it will change the original set instead of returning a new set.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set1.symmetric_difference_update(set2)
print(set1)
```

####Python Set isdisjoint() Method
The isdisjoint() method returns True if none of the items are present in both sets, otherwise it returns False.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.isdisjoint(set2)
print(set3)
```

####Python Set issubset() Method
The issubset() method returns True if all items in the set exists in the specified set, otherwise it returns False.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.issubset(set2)
print(set3)
```

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1 <= set2
print(set3)
```

####Python Set issuperset() Method
The issuperset() method returns True if all items in the specified set exists in the original set, otherwise it returns False.

```
set1 = {"clever","studies","classes"}
set2 = {"are","going","on"}
set3 = set1.issuperset(set2)
print(set3)
```
