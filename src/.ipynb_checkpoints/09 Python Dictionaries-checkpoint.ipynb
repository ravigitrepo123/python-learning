####Python Dictionaries

Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered*, changeable and do not allow duplicates.

As of Python version 3.7, dictionaries are ordered. In Python 3.6 and earlier, dictionaries are unordered.

```python
#Dictionaries are written with curly brackets, and have keys and values:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
print(dict1)
```

####Dictionary Items
Dictionary items are ordered, changeable, and do not allow duplicates.

Dictionary items are presented in key:value pairs, and can be referred to by using the key name.

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
print(dict1["classes"])
```

####Changeable
Dictionaries are changeable, meaning that we can change, add or remove items after the dictionary has been created.

####Duplicates Not Allowed
Dictionaries cannot have two items with the same key:

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on",
         "classes" : "in progress"}
print(dict1)
```

####Dictionary Length
To determine how many items a dictionary has, use the len() function:

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
print(len(dict1))
```

####Dictionary Items - Data Types
The values in dictionary items can be of any data type:

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on",
         "year"    : 2024,
         "clever"  : ["studies","classes","are","going on"]}
print(dict1)    
print(type(dict1))     
```

####The dict() Constructor
It is also possible to use the dict() constructor to make a dictionary.

```python
dict1 = dict(name = "master", age = 41, country = "India")
print(dict1)
```

####Python - Access Dictionary Items
You can access the items of a dictionary by referring to its key name, inside square brackets:

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x    = dict1["going"]
print(x)
```

```python
#There is also a method called get() that will give you the same result:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x    = dict1.get("going")
print(x)
```

```python
#The keys() method will return a list of all the keys in the dictionary
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x    = dict1.keys()
print(x)
```

```python
#Add a new item to the original dictionary, and see that the keys list gets updated as well
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x    = dict1.keys()
print(x) # before change
dict1["currently"] = "in progress" 
print(x) # after change
```

```python
#The values() method will return a list of all the values in the dictionary.
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x    = dict1.values()
print(x)
```

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
if "going" in dict1:
    print("Yes,'going' is one of the keys in this dict1")
```

####Python - Change Dictionary Items
You can change the value of a specific item by referring to its key name:

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict1["classes"] = "in progress"
print(dict1)
```

#### Update Dictionary
The update() method will update the dictionary with the items from the given argument.

The argument must be a dictionary, or an iterable object with key:value pairs.

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict1.update({"classes":"are in progress"})
print(dict1)
```

####Python - Remove Dictionary Items
There are several methods to remove items from a dictionary:

```python
#The pop() method removes the item with the specified key name:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict1.pop("classes")
print(dict1)
```

```python
#The popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead):
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict1.popitem()
print(dict1)
```

```python
#The del keyword removes the item with the specified key name:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
del dict1["classes"]
print("dict1")
```

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
del dict1
print(dict1)
#this will cause an error because "thisdict" no longer exists.
```

```python
#The clear() method empties the dictionary:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
x = dict1.clear()
print(x)
```

#### Python - Copy Dictionaries

```python
#Make a copy of a dictionary with the copy() method:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict2 = dict1.copy()
print(dict2)
```

```python
#Make a copy of a dictionary with the dict() function:
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict2 = dict(dict1)
print(dict2)
```

####Python - Nested Dictionaries
A dictionary can contain dictionaries, this is called nested dictionaries.

```python
family = {
  "child1" : {
    "name" : "Aditya",
    "year" : 1998
  },
  "child2" : {
    "name" : "Suresh",
    "year" : 2002
  },
  "child3" : {
    "name" : "Hemanth",
    "year" : 2006
  }
}
print(family)
```

```python
child1 = {
             "name" : "Aditya",
             "year" : 1998
         }
child2 = {
             "name" : "Suresh",
             "year" : 2002
         }
child3 = {
            "name" : "Hemanth",
            "year" : 2006
         }
family = {
           "child1" : child1,
           "child2" : child2,
           "child3" : child3
         }
print(family)
```

####Python Dictionary fromkeys() Method.
The fromkeys() method returns a dictionary with the specified keys and the specified value.

```python
x = ('key1', 'key2', 'key3')
y = 0
dict1 = dict.fromkeys(x, y)
print(dict1)
```

####Python Dictionary setdefault() Method
The setdefault() method returns the value of the item with the specified key.


```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict2 = dict1.setdefault("classes","in progress")
print(dict2)
```

```python
dict1 = {"clever"  : "studies",
         "classes" : "are",
         "going"   : "on"}
dict2 = dict1.setdefault("currently","in progress")
print(dict2)
```

# Find the occurances count of each character of a string

```python
s = 'Apple'
d = {}
for i in s:
    if i in d:
        # for second and next occurances increment by 1 
        d[i] = d[i]+1
    else:
        # for first occurance assign value 1 for that character
        d[i] = 1    
print(d)
```

```python
#  Dictionary.items -> key,value as tuples in a list usefull to sort on keys and values, to iterate over dict and process keys and values
```

```python
data = {
    'apple': 10,
    'banana': 5,
    'cherry': 15
}

print(data.items())
print(type(data.items()))
```

```python
#   Sort the dictionary based on keys or values
```

```python
data = {
    'apple': 10,
    'banana': 5,
    'cherry': 15,
    'apples': 1,
    'bananas': 4,
    'cherrys': 70
}

sorted_items = sorted(data.items(), key=lambda x: x[1])
print(sorted_items) #sort based of values ascending by default reverse = False

sorted_items = sorted(data.items(), key=lambda x: x[1], reverse=True)
print(sorted_items) #sort based on values descending

sorted_items = sorted(data.items(), key=lambda x:-x[1])
print(sorted_items) #sort based on values descending

sorted_items = sorted(data.items(), key=lambda x: x[0])
print(sorted_items) # sort based on keys ascending

sorted_items = sorted(data.items(), key=lambda x: x[0], reverse=True)
print(sorted_items) #sort based on keys descending

```

```python
# sort by keys and values , only first letter of key
```

```python
sorted_items = sorted(data.items(), key=lambda x: (x[0],x[1]), reverse=True)
print(sorted_items) # descending order of key then desc of value
sorted_items = sorted(data.items(), key=lambda x: (x[0],x[1]))
print(sorted_items) #ascending order of key, then value
sorted_items = sorted(data.items(), key=lambda x: (x[0][0],x[1]))
print(sorted_items) # ascending order of first letter of key then ascending order of value
```

```python
# convert the dict.items back to dictionary
```

Find the top three most common characters in the string.

Print the three most common characters along with their occurrence count.
Sort in descending order of occurrence count.
If the occurrence count is the same, sort the characters in alphabetical order.

```python
if __name__ == '__main__':
    s = input()
    d = {}
    for i in s:
        if i in d:
            d[i] = d[i]+1
        else:
            # for first occurance assign value 1 for that character
            d[i] = 1    
    #print(d)
    
    # sort desc by value then asc key
    sorted_items = sorted(d.items(), key = lambda x: (-x[1],x[0]))
    
    j = 1 
    for k,v in sorted_items:
        if j <= 3:
           print(k,v)
           j = j+1

```
