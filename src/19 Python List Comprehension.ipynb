####Python - List Comprehension

###List Comprehension
List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

Example:

Based on a list of demo, you want a new list, containing only the demo with the letter "c" in the name.

Without list comprehension you will have to write a for statement with a conditional test inside:

```python
demo = ["clever", "studies", "classes", "are", "going on"]
newlist = []

for x in demo:
    if "a" in x:
        newlist.append(x)

print(newlist)
```

```python
#With list comprehension you can do all that with only one line of code:
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist)
```

###The Syntax
- newlist = [expression for item in iterable if condition == True]
- The return value is a new list, leaving the old list unchanged.

###Condition
The condition is like a filter that only accepts the items that evaluate to True.

```python
#Only accept items that are not "apple":
demo = ["clever", "studies", "classes", "are", "going on", "apple"]

# Only accept items that are not "apple"
newlist = [x for x in demo if x != "apple"]

print(newlist)
```

The condition if x != "apple"  will return True for all elements other than "apple", making the new list contain all fruits except "apple".

The condition is optional and can be omitted:

```python
#With no if statement:
demo = ["clever", "studies", "classes", "are", "going on"]
newlist = [x for x in demo]
```

###Iterable
The iterable can be any iterable object, like a list, tuple, set etc.

```python
#You can use the range() function to create an iterable:
newlist = [x for x in range(10)]
```

```python
#Accept only numbers lower than 5:
newlist = [x for x in range(10) if x < 5]
```

###Expression
The expression is the current item in the iteration, but it is also the outcome, which you can manipulate before it ends up like a list item in the new list:

```python
demo = ["clever", "studies", "classes", "are", "going on"]

# Set the values in the new list to upper case
newlist = [x.upper() for x in demo]

print(newlist)
```

```python
#Set all values in the neThe expression can also contain conditions, not like a filter, but as a way to manipulate the outcome:
#Return "clever" instead of "studies":
demo = ["clever", "studies", "classes", "are", "going on"]

# Replace "clever" with "studies"
newlist = [x if x != "clever" else "studies" for x in demo]

print(newlist)
```

Can list comprehension replace map() and filter()? Give an example.

```python
numbers = [1, 2, 3, 4, 5, 6]

# filter even numbers, then square them
result = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, numbers)))

print(result)  # Output: [4, 16, 36]
```

Flatten a nested list using list comprehension.

Example: [[1,2],[3,4],[5,6]] → [1,2,3,4,5,6]

```python
nested_list = [[1, 2], [3, 4], [5, 6]]
fl = [x for sub_list in nested_list for x in sub_list]  # nested list comprehension
print(fl)
```

Generate a dictionary with list comprehension. example {(1:1),(2:4),(3:9),(4:16)}

Data cleanup remove None , blanks , empty space data

```python
names = ["Ravi", "", None, "Anu", " ", "Krish"]

clean_names = [n for n in names if n and n.strip()]

print(clean_names)
# if n → removes None and empty strings ("")
# n.strip() → removes leading/trailing spaces
# if n.strip() → ensures entries like " " (only spaces) are filtered out 
```
