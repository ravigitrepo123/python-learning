###Python RegEx
A RegEx, or Regular Expression, is a sequence of characters that forms a search pattern.

RegEx can be used to check if a string contains the specified search pattern.

####RegEx Module
Python has a built-in package called re, which can be used to work with Regular Expressions.

Import the re module:

```python
import re
```

####RegEx in Python
When you have imported the re module, you can start using regular expressions:

####The findall() Function
The findall() function returns a list containing all matches.

```python
#The findall() function returns a list containing all matches.
import re
x = "clever studies classes are going on"
y = re.findall("go",x)
print(y)
```

```python
#Return an empty list if no match was found:
import re
x = "clever studies classes are going on"
y = re.findall("oi",x)
print(y)
```

####Metacharacters
Metacharacters are characters with a special meaning:

```python
#   [] A set of characters
import re
x = "clever studies classes are going on"
y = re.findall("[a-k]",x)
print(y)
##Finds all lower case characters alphabetically between "a" and "k"
```

```python
#        \	Signals a special sequence (can also be used to escape special characters)
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall(r"\d",x)
print(y)
##Find all digit characters:
```

```python
#   .	Any character (except newline character)
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("go..g",x)
print(y)
#Search for a sequence that starts with "go", followed by two (any) characters, and an "o":
```

```python
#          ^	Starts with
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("^from",x)
print(y)
##Check if the string starts with 'from':
```

```python
#             $   	Ends with
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("onwards$",x)
print(y)
##Check if the string ends with 'onwards':
```

```python
#       *	  Zero or more occurrences
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("go*g",x)
print(y)
#Search for a sequence that starts with "go", followed by 0 or more  (any) characters, and an "g":
```

```python
#        +  	One or more occurrences
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("go+g",x)
print(y)
#Search for a sequence that starts with "go", followed by 1 or more  (any) characters, and an "g":
```

```python
#  ?	Zero or one occurrences
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("go.?g",x)
print(y)
#Search for a sequence that starts with "go", followed by 0 or 1  (any) character, and an "g":
```

```python
#    {}	Exactly the specified number of occurrence
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("go{2}g",x)
print(y)
##Search for a sequence that starts with "go", followed excactly 2 (any) characters, and an "g":
```

```python
#  |	Either or
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("going|now",x)
print(y)
#Check if the string contains either "going" or "now":
```

####Special Sequences
A special sequence is a \ followed by one of the characters in the examples given below, and has a special meaning:

```python
#          \A	  Returns a match if the specified characters are at the beginning of the string.
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\Aclever",x)
print(y)
#Checks if the string starts with "clever":
```

```python
#    \b	Returns a match where the specified characters are at the beginning or at the end of a word.

#(the "r" in the beginning is making sure that the string is being treated as a "raw string")
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall(r"\bonwards",x)
print(y)
##Check if "onwards" is present at the beginning of a WORD:
```

```python
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall(r"onwards\b",x)
print(y)
##Check if "onwards" is present at the end of a WORD:
```

```python
#\B	Returns a match where the specified characters are present, but NOT at the beginning (or at the end) of a word
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall(r"\Bonwards",x)
print(y)
#Check if "onwards" is present, but NOT at the beginning of a word:
```

```python
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall(r"are\B",x)
print(y)
##Check if "ain" is present, but NOT at the end of a word:
```

```python
#      \d	Returns a match where the string contains digits (numbers from 0-9)
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\d",x)
print(y)
##Check if the string contains any digits (numbers from 0-9):
```

```python
#       \D	      Returns a match where the string DOES NOT contain digits
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\D",x)
print(y)
##Return a match at every no-digit character:
```

```python
#          \s	   Returns a match where the string contains a white space character
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\s",x)
print(y)
#Return a match at every white-space character:
```

```python
# \S	Returns a match where the string DOES NOT contain a white space character
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\S",x)
print(y)
#Return a match at every NON white-space character:
```

```python
#    \w	     Returns a match where the string contains any word characters (characters from a to Z, digits from 0-9, and the underscore _ character)
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\w",x)
print(y)
#Return a match at every word character (characters from a to Z, digits from 0-9, and the underscore _ character)
```

```python
#\W	Returns a match where the string DOES NOT contain any word characters
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("\W",x)
print(y)
##Return a match at every NON word character (characters NOT between a and Z. Like "!", "?" white-space etc.):
```

```python
#          \Z	Returns a match if the specified characters are at the end of the string
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("onwards\Z",x)
print(y)
##Check if the string ends with "onwards":
```

####Sets
A set is a set of characters inside a pair of square brackets [] with a special meaning:

```python
#[abc]	Returns a match where one of the specified characters (a, b, or c) is present
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[abc]",x)
print(y)
##Check if the string has any a, b, or c characters:
```

```python
#[a-e]	Returns a match for any lower case character, alphabetically between a and e
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[a-e]",x)
print(y)
```

```python
#[^abc]	Returns a match for any character EXCEPT a, b, and c
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[^abc]",x)
print(y)
```

```python
#[23678]	Returns a match where any of the specified digits (2, 3, 6,7 or 8) are present
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[23678]",x)
print(y)
```

```python
#[0-9]	Returns a match for any digit between 0 and 9
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[0-9]",x)
print(y)
```

```python
#[0-6][0-9]	Returns a match for any two-digit numbers from 00 and 69
import re
x = "clever studies classes are going on from 8.30 AM onwards"
y = re.findall("[0-6][0-9]",x)
print(y)
```

```python
#[a-zA-Z]	Returns a match for any character alphabetically between a and z, lower case OR upper case
import re
x = "clever studies Classes Are going on from 8.30 AM onwards"
y = re.findall("[a-zA-Z]",x)
print(y)
##Check if the string has any characters from a to z lower case, and A to Z upper case:
```

```python
#[+]	In sets, +, *, ., |, (), $,{} has no special meaning, so [+] means: return a match for any + character in the string
import re
x = "clever studies Classes Are going on from 8.30 AM onwards"
y = re.findall("[+]",x)
print(y)
#Check if the string has any + characters:
```

####RegEx Functions
The re module offers a set of functions that allows us to search a string for a match:

####The search() Function
The search() function searches the string for a match, and returns a Match object if there is a match.

If there is more than one match, __only the first occurrence__ of the match will be returned:

```python
import re
x = "clever studies classes are going on"
y = re.search("\s",x)
print(y)
```

```python
#If no matches are found, the value None is returned:
import re
x = "clever studies classes are going on"
y = re.search("ai",x)
print(y)
```

####Match Object
A Match Object is an object containing information about the search and the result.

Note: If there is no match, the value None will be returned, instead of the Match Object.

```python
import re
x = "clever studies classes are going on"
y = re.search("es",x)
print(y)
#this will print an object
```

The Match object has properties and methods used to retrieve information about the search, and the result:

.span() returns a tuple containing the start-, and end positions of the match.

.string returns the string passed into the function.

.group() returns the part of the string where there was a match.
