### Memory locations of variables

```python
a = [1, 2, 3]
b = a
a += [4, 5]
print(a)
print(b)
b += [5,6]
print(a)
print(b)
# Memory locations will be same for both varaibles, when we assign it using = 
print(id(a))
print(id(b))
c=a[:]
# When you copy with index Memory locations of new varable will be different
print(c)
print(id(c))
# copy menthod also use different memory location and copies only values
d = a.copy()
print(d)
print(id(d))
# list menthod also use different memory location and copies only value
e = list(a)
print(e)
print(id(e))
```

###GLOBAL VARIABLES

Variables that are created outside of a function (as in all of the examples in the previous pages) are known as global variables.

Global variables can be used by everyone, both inside of functions and outside

```python
x = "HELLO WORLD!"

def myfunc():
  print("HELLO WORLD!" + x)

myfunc()
```

If you create a variable with the same name inside a function, this variable will be local, and can only be used inside the function. The global variable with the same name will remain as it was, global and with the original value.

```python
x = "HELLO WORLD!"
def myfunc():
  x = "HELLO WORLD!"
  print("MY FIRST PROGRAM IS: " + x)
myfunc()
print("MY FIRST PROGRAM IS:" + x)
```

###The global Keyword
Normally, when you create a variable inside a function, that variable is local, and can only be used inside that function.

To create a global variable inside a function, you can use the global keyword.

```python
def myfunc():
  global x
  x = "HELLO WORLD"

myfunc()

print("MY FIRST PROGRAM IS:" + x)
```

#####To change the value of a global variable inside a function, refer to the variable by using the global keyword:

```python
x = "HELLO"

def myfunc():
  global x
  x = "WORLD"

myfunc()

print("MY FIRST PROGRAM IS" + x)
```
